```python
df_list=[]
for i,file in enumerate(list_files):
    print(i, file, sep='')
    df= pd.read_excel(list_files[i],sheet_name=1)
    df_list.append(df)
my_df = pd.concat(df_list, axis=0)
```

```python
# Saving Stacked Data Frame
stacked_df_path = path+'Stacked_Excel_File_MA_AR_New\\'
print(stacked_df_path)
# Create the directory 
# 'GeeksForGeeks' in 
# '/home / User / Documents' 
try: 
    os.mkdir(stacked_df_path) 
    os.chdir(stacked_df_path)
    df_dia_ma_ar_cir.to_csv('Stacked_MA_AR_Cir'+"_"+my_b4c_type[0:9]+".csv")
except OSError as error: 
    print(error)  
```

```python
hist_path = path+"PLOTS\\Histograms_NEW"
# Create the directory 
# 'GeeksForGeeks' in 
# '/home / User / Documents' 
try: 
    os.mkdir(hist_path) 
    os.chdir(hist_path)
    for col in df_dia_ma_ar_cir.columns:
        print(col)
        sns.histplot(df_dia_ma_ar_cir[col])
        plt.title('Histogram:'+" "+str(col))
        plt.savefig(fname= my_b4c_type[0:9]+"_"+str(col)+"_"+'Histogram')
        plt.show()

except OSError as error: 
    print(error)  
```

```python
dist_path = path+"PLOTS\\Distribution_Plots_NEW"
# Create the directory 
# 'GeeksForGeeks' in 
# '/home / User / Documents' 
try: 
    os.mkdir(dist_path) 
    os.chdir(dist_path)
    for col in df_dia_ma_ar_cir.columns:
        sns.distplot(df_dia_ma_ar_cir[col])
        plt.title('Distribution_Plot:'+" "+str(col))
        plt.savefig(fname= my_b4c_type[0:9]+"_"+str(col)+"_"+'Distribution_Plot')
        plt.show()
except OSError as error: 
    print(error)  
```
